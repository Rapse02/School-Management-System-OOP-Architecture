```python
class Persona:
    """Clase que modela una persona, sus atributos y métodos básicos.
    
        Atributos:
        __nombre: string - Nombre de la persona
        __apellido: string - Apellido de la persona
        __edad: int - Edad de la persona
        __sexo: string – Corresponde al sexo de la persona

        Metodos:
        registrar_datos(nombre, apellido, edad) - Permite registrar los datos de la persona
        get_nombre() - Devuelve el nombre completo de la persona - nombre + apellido
        ver_datos_persona() - Imprime los datos de la persona
        """
    def __init__(self):
        self.__nombre = ""
        self.__apellido = ""
        self.__edad = 0
        self.__sexo = ""

    def registrar_datos(self):
        """"Metodo que permite registrar los datos de una persona"""
        self.__nombre = input("Ingrese el nombre de la persona: ")
        self.__apellido = input("Ingrese el apellido de la persona: ")
        self.__edad = int(input("Ingrese la edad de la persona: "))
        self.__sexo = input("Ingrese el sexo de la persona: ")

    def modificar_datos_persona(self, opcion):
        """Metodo para actualizar datos de la persona desde las subclases.
            Solicita un argumento opcion-int- que debe ser proporcionado desde la subclase.
            Opciones: 1 = nombre, 2 = apellido, 3 = edad, 4 = sexo."""
        if opcion == 1:
            self.__nombre = input("Ingrese el nombre de la persona: ")

        elif opcion == 2:
            self.__apellido = input("Ingrese el apellido de la persona: ")

        elif opcion == 3:
            self.__edad = int(input("Ingrese la edad de la persona: "))

        elif opcion == 4:
            self.__sexo = input("Ingrese el sexo de la persona: ")

        else:
            pass
        

    def get_nombre(self):
        """"Metodo para devolver el nombre completo de la persona"""
        return self.__nombre, self.__apellido

    def ver_datos_persona(self):
        """Metodo que imprimer todos los datos de la persona"""
        print (f"Nombre: {self.__nombre}\nApellido: {self.__apellido}\nEdad: {self.__edad}\nSexo: {self.__sexo}\n")
```

```python
class Alumno(Persona):
    """Subclase de la superclase Persona. Modela el objeto alumno, para una escuela.

        Atributos heredados:
                __nombre: string - Nombre de la persona
                __apellido: string - Apellido de la persona
                __edad: int - Edad de la persona
                __sexo: string – Corresponde al sexo de la persona
        
        Atributos:
                __num_control_escolar: string – Corresponde al número de control escolar asignado al alumno por la escuela
                __semestre_alumno: int – Corresponde al semestre que cursa el alumno actualmente
                __materias_cursa_alumno: list – Lista de objeto_materia que cursa el alumno
        
        Métodos:
                ver_datos_alumno: Imprime toda la información del alumno
                modificar_datos_alumno: – Permite modificar la información del alumno
                inscribir_materias_alumno: – Permite agregar materias que cursa el alumno
                baja_materia_alumno: – Permite eliminar materias de la lista del alumno"""
    
    def __init__(self):
        super().__init__()
        self.__num_control_escolar = ""  
        self.__semestre_alumno = ""
        self.__materias_cursa_alumno = [] 


    
    def añadir_información_alumno (self):
        super().registrar_datos()
        self.__num_control_escolar = input("Ingrese el número de control escolar del alumno: ")
        self.__semestre_alumno = int(input("Ingrese el semestre que cursa el alumno: "))
    
    def ver_datos_alumno (self):
        super().ver_datos_persona()
        print (f"Numero de control: {self.__num_control_escolar}\nSemestre del alumno: {self.__semestre_alumno}\n")
        print("\nMaterias inscritas:")
        for materia in self.__materias_cursa_alumno:
            print (materia.ver_id())
        print("\nSe imprimieron los datos del alumno\n")

    def ver_num_control_escolar(self):
        return self.__num_control_escolar
    
    def modificar_datos_alumno(self):
        """Metodo para modificar datos del alumno.
        Usa el metodo de la superclase para modiciar datos de la persona."""
        opcion = int(input("Seleccione el dato a modificar: 1. Nombre\n2. Apellido\n3. Edad\n4. Sexo\n5. Numero de control\n6. Semestre"))

        super().modificar_datos_persona(opcion)
        
        if opcion == 5:
            self.__num_control_escolar = input("Ingrese el número de control escolar del alumno: ")
        
        elif opcion == 6:
            self.__semestre_alumno = int(input("Ingrese el semestre que cursa el alumno: "))

        else:
            print("\nOpción no valida.")
        print("\nSe modificaron los datos del alumno")

    def inscribir_materias_alumno (self, obj_materia):
        self.__materias_cursa_alumno.append(obj_materia)
        print (self.__materias_cursa_alumno)        
        print("Se inscribieron las materias correctamente")

    def baja_materias_alumno (self, obj_materia):
        self.__materias_cursa_alumno.remove(obj_materia)
        print("Se dieron de baja las materias exitosamente")
```

```python
class Profesor(Persona):
    """Clase que modela el objeto profesor, para una escuela. Es una subclase de la superclase Persona.

        Atributos heredados:
                __nombre: string - Nombre de la persona
                __apellido: string - Apellido de la persona
                __edad: int - Edad de la persona
                __sexo: string – Corresponde al sexo de la persona
        
        Atributos:
                __titulo_profesor: int – Corresponde al título o profesión del profesor
                __materias_impartidas: list – Corresponde al conjunto de objetos_materias que imparte el profesor
                __num_cedula_profesional: string – Corresponde a la cedula profesional del profesor
                __num_profesor_control: string – Corresponde al número de identificación escolar asignado al profesor
                
        Métodos:
                ver_datos_profesor () – Imprime los datos del profesor
                modificar_datos_profesor () – Permite modificar los datos del profesor
                asignar_materia () – Permite asignar las materias que impartirá el profesor
                eliminar_materia () – Permite eliminar materias que impartirá el profesor"""
    
    def __init__(self):
        super().__init__()
        self.__titulo_profesor = ""
        self.__materias_impartidas = []
        self.__num_cedula_profesional = ""
        self.__num_profesor_control = ""

    def añadir_datos_profesor (self):
        super().registrar_datos()
        self.__titulo_profesor = input("Ingrese el titulo o profesión del profesor: ")
        self.__num_cedula_profesional = input("Ingrese el número de cedula profesional del profesor: ")
        self.__num_profesor_control = input("Ingrese el número de profesor asignado: ")

    def ver_datos_profesor (self):
        super().ver_datos_persona()
        print(f"\nTitulo: {self.__titulo_profesor}\nCedula profesional: {self.__num_cedula_profesional}\nNumero de profesor: {self.__num_profesor_control}")
        print("\nMaterias impartidas:")
        for materia in self.__materias_impartidas:
            print (materia.ver_id())
        print("Se imprimieron los datos del profesor")

    def ver_num_control (self):
        return self.__num_profesor_control

    def modificar_datos_profesor(self):
        """Metodo para actualizar datos del profesor.
        Utiliza el méotodo modificar de la superclase Persona."""
        opcion = input("Seleccione el dato a modificar: 1. Nombre\n2. Apellido\n3. Edad\n4. Sexo\n5. Titulo profesional\n6. Cedula profesional\n7. Numero de control")

        super().modificar_datos_persona(opcion)
        
        if opcion == 5:
            self.__titulo_profesor = input("Ingrese el titulo o profesión del profesor: ")
        
        elif opcion == 6:
            self.__num_cedula_profesional = input("Ingrese el número de cedula profesional del profesor: ")

        elif opcion == 7:
            self.__num_profesor_control = input("Ingrese el número de profesor asignado: ")

        else:
            print("\nOpción no valida.")
        print("\nSe modificaron los datos del profesor")
        
    def asignar_materia (self, obj_materia):
        self.__materias_impartidas.append(obj_materia)
        print("Se asigno la materia al profesor exitosamente")

    def eliminar_materia (self, obj_materia):
        self.__materias_impartidas.remove(obj_materia)
        print("Se elimino la materia del profesor exitosamente")
```

```python
class Materia:
    """ Clase que modela el objeto materia, para una escuela
        
        Atributos:
                __nombre_materia: string – Corresponde al nombre de la materia
                __num_creditos: int – Corresponde al número de créditos que da la materia
                __profesor_asignado: objeto_profesor que imparte la materia
                __id: ID de la materia
        
        Métodos:
                ver_datos_materia () – Imprime los datos de la materia
                actualizar_datos_materia () – Permite actualizar los datos de la materia
                asignar_profesor () – Permite asignar el profesor que impartirá la materia"""

    def __init__(self):
        self.__nombre_materia = ""
        self.__num_creditos = 0
        self.__profesor_asignado = None
        self.__id = ""


    
    def añadir_datos_materia(self):
        self.__nombre_materia = input("Ingrese el nombre de la materia: ")
        self.__num_creditos = int(input("Ingrese el número de creditos correspondientes a la materia: "))
        self.__id = input("Ingrese el id de la materia: ")

    def ver_datos_materia(self):
        print(f"\nMateria: {self.__nombre_materia}\nCreditos: {self.__num_creditos}\nProfesor asignado: {self.ver_profesor()}")
        print("Se imprimieron los datos de la materia")

    def ver_id(self):
        return self.__id
    
    def ver_profesor(self):
        if self.__profesor_asignado == None:
            print("Sin profesor asignado")
        else:
            return self.__profesor_asignado.ver_num_control()
    
    def actualizar_datos_materia(self):
        opcion = input("Ingresa el dato a modificar: 1. Nombre\n2. Número de creditos\n 3.ID")

        if opcion == 1:
            self.__nombre_materia = input("Ingrese el nombre de la materia: ")
        
        elif opcion == 2:
            self.__num_creditos = int(input("Ingrese el número de creditos correspondientes a la materia: "))

        elif opcion == 3:
            self.__id = input("Ingrese el id de la materia: ")
        
        else:
            print("\nOpcion no valida.")
        
        print("\nSe actualizaron los datos de la materia exitosamente")

    def asignar_profesor (self, obj_profesor):
        self.__profesor_asignado = obj_profesor
        print("Se asigno el profesor exitosamente")
```

```python
lista_alumnos = []

lista_materias = []

lista_profesores = []
```

```python
def añadir_alumno_a_lista():
    """Funcion para añadir un nuevo registro de alumno"""
    alumno = Alumno()
    alumno.añadir_información_alumno()
    lista_alumnos.append(alumno)
    print("Alumno añadido con éxito")
```

```python
def añadir_materia_a_lista():
    """Funcion para añadir un nuevo registro de materia"""
    materia = Materia()
    materia.añadir_datos_materia()
    lista_materias.append(materia)
    print("Materia añadida con éxito")
```

```python
def añadir_profesor_a_lista():
    """Funcion para añadir un nuevo registro de profesor"""
    profesor = Profesor()
    profesor.añadir_datos_profesor()
    lista_profesores.append(profesor)
    print("Profesor añadido con éxito")
```

```python
def buscar_registro(opcion):
    """Funcion para buscar un registro con base en el ID"""
    id_buscado = input("Ingrese el ID o número de control buscado: ")
    encontrado = False

    if opcion == 1:
        for alumno in lista_alumnos:
            if id_buscado == alumno.ver_num_control_escolar():
                alumno.ver_datos_alumno()
                encontrado = True
                return
        if encontrado == False:
            print("No se encontro el numero de control ingresado.")

    elif opcion == 2:
        for materia in lista_materias:
            if id_buscado == materia.ver_id():
                materia.ver_datos_materia()
                encontrado = True
                return
        if encontrado == False:
            print("No se encontro el Id ingresado.")

    elif opcion == 3:
        for profesor in lista_profesores:
            if id_buscado == profesor.ver_num_control():
                profesor.ver_datos_profesor()
                encontrado = True
                return
        if encontrado == False:
            print("No se encontro el numero de control ingresado.")

    else:
        print("\nOpcion no valida.")
```

```python
def visualizar_registro():
    """Menu para visualizar registros"""
    opcion = int(input("Ingrese la opcion que desea visualizar: \n1. Alumno\n2. Materia\n3. Profesor\n"))
    buscar_registro(opcion)
```
